## The data

We use **Gapminder**, a dataset of life expectancy, income (GDP per person) and population for 142
countries every five years from 1952 to 2007. It comes with the Plotly library, so there is nothing to
download.

| Column | Meaning |
|---|---|
| `country`, `continent` | Names |
| `year` | 1952, 1957, ..., 2007 |
| `lifeExp` | Life expectancy at birth, in years |
| `pop` | Population |
| `gdpPercap` | GDP per person, in inflation-adjusted U.S. dollars |

## Step 1: Import libraries and load the data

````python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

gap = px.data.gapminder()
gap.head()
````

`plt` is the usual short name for Matplotlib's plotting module. Each row of `gap` is one country in
one year.

## Step 2: A first line chart, and why it is not finished

Let us plot life expectancy in the United States over time.

````python
us = gap[gap["country"] == "United States"]

plt.plot(us["year"], us["lifeExp"])
plt.show()
````

The line is correct, but a reader would ask: *What is measured? In what units? Which country? Where
does the data come from?* An unlabeled chart is only useful to the person who made it.

## Step 3: Add a title, axis labels and a source

````python
plt.figure(figsize=(7, 4))
plt.plot(us["year"], us["lifeExp"], marker="o", color="#3b6ea5")
plt.title("Life expectancy in the United States rose by about 10 years, 1952 to 2007")
plt.xlabel("Year")
plt.ylabel("Life expectancy at birth (years)")
plt.figtext(0.01, -0.03, "Source: Gapminder", fontsize=8, color="grey")
plt.show()
````

A good title tells the reader **what to notice**, not just what is plotted. The source line lets
readers check the data.

## Step 4: Compare a few countries

When you plot several lines, add a `label` to each and show a legend.

````python
countries = ["United States", "Canada", "Mexico", "Brazil"]

plt.figure(figsize=(7, 4))
for name in countries:
    one = gap[gap["country"] == name]
    plt.plot(one["year"], one["lifeExp"], marker="o", label=name)

plt.title("Life expectancy in four countries of the Americas")
plt.xlabel("Year")
plt.ylabel("Life expectancy (years)")
plt.legend()
plt.show()
````

Four lines are easy to follow. With ten or more, the chart becomes a tangle; use small multiples
instead (see notebook 1b).

## Step 5: A misleading axis

Bar charts compare **lengths**, so the axis must start at zero. Here are the same four values twice.

````python
latest = gap[(gap["year"] == 2007) & (gap["country"].isin(countries))].sort_values("lifeExp")

fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4))

left.bar(latest["country"], latest["lifeExp"], color="#3b6ea5")
left.set_ylim(70, 82)                       # truncated axis
left.set_title("Misleading: axis starts at 70")
left.set_ylabel("Life expectancy (years)")

right.bar(latest["country"], latest["lifeExp"], color="#3b6ea5")
right.set_ylim(0, 85)                       # honest axis
right.set_title("Honest: axis starts at 0")

plt.show()
````

On the left, Canada's bar looks several times taller than Brazil's. On the right, you can see they are
only a few years apart. Line charts may start above zero (readers compare positions, not lengths), but
the axis must be clearly labeled.

## Step 6: The wrong chart type

Lines suggest a continuous change, such as over time. For separate categories, use bars.

````python
by_continent = gap[gap["year"] == 2007].groupby("continent")["lifeExp"].mean().sort_values()

fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4))

left.plot(by_continent.index, by_continent.values, marker="o")
left.set_title("Line: implies a trend from Africa to Oceania")
left.set_ylabel("Average life expectancy, 2007 (years)")

right.bar(by_continent.index, by_continent.values, color="#3b6ea5")
right.set_title("Bar: separate categories")

plt.show()
````

The line on the left suggests the continents are steps along a path. They are not; they are separate
groups. The bar chart shows that correctly.

## Step 7: Too many colors

Color should carry meaning. If every bar is the same kind of thing, use one color, and save a second
color for the one bar you want readers to notice.

````python
fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4))

left.bar(by_continent.index, by_continent.values,
         color=["red", "orange", "green", "blue", "purple"])
left.set_title("Distracting: a color per bar means nothing")

highlight = ["#3b6ea5" if name == "Americas" else "#c9c9c9" for name in by_continent.index]
right.bar(by_continent.index, by_continent.values, color=highlight)
right.set_title("Focused: one color, one highlight (Americas)")

for ax in (left, right):
    ax.set_ylabel("Average life expectancy, 2007 (years)")
plt.show()
````

## Step 8: Long labels

Country names overlap on a vertical bar chart. A **horizontal** bar chart fixes this and is easy to
sort.

````python
asia_2007 = gap[(gap["year"] == 2007) & (gap["continent"] == "Asia")]
top10 = asia_2007.sort_values("gdpPercap").tail(10)

plt.figure(figsize=(7, 5))
plt.barh(top10["country"], top10["gdpPercap"], color="#3b6ea5")
plt.title("Ten richest countries in Asia by GDP per person, 2007")
plt.xlabel("GDP per person (inflation-adjusted US dollars)")
plt.show()
````

## Good chart checklist

- [ ] A title that states the message
- [ ] Axis labels with units
- [ ] Bar charts start at zero
- [ ] The chart type matches the data (lines for time, bars for categories)
- [ ] Color is used only where it means something
- [ ] Labels are readable
- [ ] The data source is stated

## Recap

- Every explanatory chart needs a title, labeled axes with units, and a source.
- Bar charts must start at zero; line charts may not, but must be labeled.
- Use lines for change over time and bars for separate categories.
- Use color to carry meaning, not decoration; horizontal bars fix long labels.
